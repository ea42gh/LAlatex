# LAlatex Binder Demo

A compact tour of the main LAlatex display constructs in a live Julia kernel.

```julia
using LAlatex
using LaTeXStrings
using Symbolics
using LinearAlgebra
using BlockArrays

LAlatex.set_backend!(:symbolics)
LAlatex.reset_display_defaults!()
@variables x y t
```

## `l_show` and `L_show`

```julia
A = [1 2 3; 4 5 6]
display(l_show(L"A = ", A, L",\quad A^T A = ", transpose(A) * A; arraystyle=:bmatrix))
print(L_show(L"A = ", A; inline=false, arraystyle=:bmatrix))
```

## Vectors, Tuples, and Separators

```julia
v = [1, 2, 3]
row = [4 5 6]
display(l_show(L"v = ", v, L",\quad r = ", row; arraystyle=:bmatrix))
display(l_show("Tuple with custom separator: ", (:alpha, :beta, 2//3); separator=L";\quad "))
```

## Mixed Numeric and Symbolic Matrices

```julia
M = mixed_matrix((1//2, x), ((1 + im)//3, 2y))
N = @mixed_matrix [1//3 x + y; x*y 2]
display(l_show(L"M = ", M, L",\quad N = ", N; arraystyle=:bmatrix))
```

## `set(...)`: Finite Sets and Local Options

```julia
display(l_show(L"B = ", set(L"e_1", L"e_2", L"e_3"; separator=L",\;")))
display(l_show("Matrix set: ", set([1, 0], [0, 1]; arraystyle=:bmatrix, separator=L",\quad ")))
```

## `set(...; such_that=...)`: Set-Builder Notation

```julia
display(l_show(L"S = ", set(x; such_that=(L"x > 0", L"x < 1"), separator=L",\;")))
display(l_show(L"T = ", set((value=x, color=:blue); such_that=((value=L"x \ne 0", color=:red),), such_that_separator=L":")))
```

## `lc(...)`: Linear Combinations

```julia
coeffs = [1, -2, 0, 3]
vectors = [[1, 0], [0, 1], [1, 1], [2, -1]]
display(l_show(L"u = ", lc(coeffs, vectors; sign_policy=:signed, omit_one=true, drop_zero=true); arraystyle=:bmatrix))
display(l_show("Keep zeros: ", lc(coeffs, vectors; drop_zero=false); arraystyle=:bmatrix))
```

## `cases(...)`: Piecewise Displays

```julia
display(l_show(
    L"T(v) = ",
    cases(
        [x, 0] => L"v \in \operatorname{span}\{e_1\}",
        ([0, y], "otherwise"),
    );
    arraystyle=:bmatrix,
))
```

## `aligned(...)`: Derivations and Equation Chains

```julia
display(l_show(aligned(
    L"Ax" => L"b",
    (L"x", L"\in", L"\mathcal{N}(A)"),
    L"\dim\mathcal{N}(A)" => L"n - \operatorname{rank}(A)",
)))
```

## Tags and Labels

```julia
display(l_show(
    aligned(L"Ax" => L"b", (L"x", L"\in", L"\mathcal{N}(A)"));
    inline=false,
    tag="1",
    label="eq:binder-demo",
))
```

## Symbolic Display Options

```julia
expr = x^2 + x*y + x + y
display(l_show(L"expanded: ", (x + y)^2; symopts=(expand=true,)))
display(l_show(L"factored: ", x^2 + x*y; symopts=(factor=true,)))
display(l_show(L"collected: ", expr; symopts=(collect=x,)))
```

## Denominator Factoring

```julia
R = [1//2 1//3; 2//3 3//4]
den, scaled = factor_out_denominator(R)
display(l_show(L"R = ", R, L",\quad d = ", den, L",\quad dR = ", scaled; arraystyle=:bmatrix))
display(l_show("unfactored: ", R; factor_out=false, arraystyle=:bmatrix))
```

## Block Matrices and Element Styling

```julia
B = BlockArray([1 2 4; 3 4 5], [1, 1], [2, 1])

function color_odd_entries(value, i, j, latex)
    return isodd(value) ? "\\textcolor{red}{$latex}" : latex
end

display(l_show(L"B = ", B; per_element_style=color_odd_entries))
```

## Display Defaults

```julia
with_display_defaults(arraystyle=:bmatrix, separator=L";\quad ") do
    display(l_show("defaults in scope: ", [1 2; 3 4], (1, 2, 3)))
end

display(l_show("defaults restored: ", [1 2; 3 4]))
```

## Formatting Helpers

```julia
numbers = [pi 0.125; exp(1) 2//3]
display(l_show(L"rounded: ", numbers; number_formatter=x -> round_value(x, 2), ))
display(l_show(L"percent: ", 0.125; number_formatter=percentage_formatter))
```
